```python
num1=int(input("Enter the Num 1 : "))
num2=int(input("Enter the Num 2 : "))

process=input("Enter the Value + , -, * , / , ** ")


match(process):
  case "+":
    print(num1+num2)
  case "-":
    print(num1-num2)
  case "*":
    print(num1*num2)
  case "/":
    print(num1/num2)
  case "**":
    print(num1**num2)
  case _:
    print("Your are Enter Wrong Value. Please Enter +,-,*,/,**")
```
